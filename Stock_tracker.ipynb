```python
Available_stocks = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "AMZN": 175
}

total = 0
records = []


print("Available stocks: AAPL, TSLA, GOOGL, AMZN")
print("Type 'done' when finished")
while True:
    stock = input("Enter stock: ").upper()
    if stock == "DONE":
        break
    if stock in Available_stocks:
        quantity = int(input("Enter quantity: "))
        cost = Available_stocks[stock] * quantity
        total = total + cost
        records.append(f"{stock} x {quantity} = {cost}")
        print(f"Added: {stock} x {quantity} = {cost}")
    else:
        print("Stock not found. Try again.")

print("Total Investment:", total)

choice = input("Save to file? (yes/no): ").lower()

if choice == "yes":
    with open("investment.txt", "w") as file:
        file.write("Total Investment: " + str(total) )
    print("Saved to investment.txt!")
```

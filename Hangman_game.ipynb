```python

import random
words = ["apple", "mango", "banana", "grapes", "lemon"]
word = random.choice(words)
correct_word = ["_"] * len(word)
count = 0
total_guess_allowed = 6
while count < total_guess_allowed and "_" in correct_word:
    print(" ".join(correct_word))
    letter = input("Guess a letter: ").lower()
    if letter in word:
        for i in range(len(word)):
            if word[i] == letter:
                correct_word[i] = letter
        print("Correct guess!")
    else:
        print("Wrong guess!")
        count += 1
        print("Remaining chances:", total_guess_allowed - count)
if "_" not in correct_word:
    print("Congratulations! You won!")
    print("Word:", word)
else:
    print("You lose.Game Over!")
    print("Correct word was:", word)
```
